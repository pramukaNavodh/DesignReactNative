```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
```

```python
df = pd.read_csv("/content/Iris Dataset.csv")
```

```python
#remove unnessacary data from colums
df.columns = df.columns.str.strip()
```

```python
print(df.shape)
```

```python
print(df["class"].value_counts())
```

```python
print(df.head)
```

```python
#visualization
sns.FacetGrid(df, hue="class").map(plt.scatter,'sepal_length', 'sepal_width').add_legend()
```

```python
#visualization
sns.FacetGrid(df, hue="class").map(plt.scatter,'petal_length', 'petal_width').add_legend()
```

```python
#seperate input features and target
x = df[["sepal_length", "sepal_width", "petal_length", "petal_width"]]
y = df["class"]
```

```python
print(y.shape)
```

```python
#encode class labels
encoder = LabelEncoder()
y1 = encoder.fit_transform(y)
```

```python
print(y1)
```

splitting data in to training and test

```python
x_train, x_test, y_train, y_test = train_test_split(
    x, y1, test_size=0.2, random_state = 42)
```

```python
#scaling inputs features

scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

Building a neural network

```python
model = Sequential([
    Dense(8, activation="relu", input_shape=(4,)),
    Dense(3, activation="softmax")
])
```

```python
#compile model
model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
```

```python
#training the model
model.fit(x_train, y_train, epochs=100, batch_size=8)
```

```python
#evaluate model

loss, accuracy = model.evaluate(x_test, y_test)
print("Test accuracy:",round(accuracy * 100,2),"%")
```

prediction

```python
probabilities = model.predict(x_test)
predictions = np.argmax(probabilities, axis=1)
print("Actual Classes",y_test)
print("predicted Classes", predictions)
```
